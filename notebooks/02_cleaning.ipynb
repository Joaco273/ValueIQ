```python
import pandas as pd
import numpy as np

df = pd.read_csv('../data/raw/AmesHousing.csv')
print("Original shape:", df.shape)
```

```python
cols_to_drop = ['Pool QC', 'Misc Feature', 'Alley', 'Fence']
df.drop(columns=cols_to_drop, inplace=True)
print("After dropping high-missing cols:", df.shape)
```

```python
none_fill_cols = [
    'Mas Vnr Type', 'Fireplace Qu',
    'Garage Qual', 'Garage Cond', 'Garage Finish', 'Garage Type',
    'Bsmt Exposure', 'BsmtFin Type 2', 'Bsmt Qual', 'Bsmt Cond', 'BsmtFin Type 1'
]

for col in none_fill_cols:
    df[col] = df[col].fillna('No')
```

```python
df['Garage Yr Blt'] = df['Garage Yr Blt'].fillna(df['Year Built'])
```

```python
median_fill_cols = [
    'Lot Frontage', 'Mas Vnr Area', 'Bsmt Full Bath', 'Bsmt Half Bath',
    'Total Bsmt SF', 'BsmtFin SF 1', 'BsmtFin SF 2',
    'Garage Area', 'Garage Cars', 'Bsmt Unf SF'
]

for col in median_fill_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
df['Electrical'] = df['Electrical'].fillna(df['Electrical'].mode()[0])
```

```python
df = df[~((df['Gr Liv Area'] > 4000) & (df['SalePrice'] < 200000))]
print("After outlier removal:", df.shape)
```

```python
remaining_nulls = df.isnull().sum().sum()
print(f"Remaining nulls: {remaining_nulls}")

# Should print 0 — if not, find what's left
if remaining_nulls > 0:
    print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
df.to_csv('../data/processed/cleaned.csv', index=False)

# Verify the saved file
clean = pd.read_csv('../data/processed/cleaned.csv')
print("Final clean shape:", clean.shape)
print("Nulls in saved file:", clean.isnull().sum().sum())
```
